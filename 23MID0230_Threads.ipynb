```python
#TASK 1 - SEQUENTIAL EXECUTION OF THREADS

import threading
import time

start=time.perf_counter()

def do_something():
    print("Sleeping 1 sec")
    time.sleep(1)
    print("Done sleeping")
    
t1=threading.Thread(target=do_something)
t2=threading.Thread(target=do_something)

t1.start()
t2.start()

t1.join()
t2.join()

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')

```

```python
#TASK 2 - CREATING A LIST OF THREADS IN FOR LOOP

import threading
import time

start=time.perf_counter()

def do_something():
    print("Sleeping 1 sec")
    time.sleep(1)
    print("Done sleeping")

l1=[]

for i in range(20):
    t=threading.Thread(target=do_something)
    t.start()
    l1.append(t)

for i in range(20):
     t.join()

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')


    
    
```

```python
#TASK 3 - PASSING ARGUMENTS TO THREADS IN LOOP

import threading
import time

start=time.perf_counter()

def do_something(i):
    print(f"Thread:{i} Sleeping 1 sec")
    time.sleep(1) 
    print("Done sleeping") 

l=[]

for i in range(10):
    t=threading.Thread(target=do_something,args=[i])
    t.start()
    l.append(t)
for t in l:
    t.join()

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')

```

```python
#TASK 4 - IMPORT CONCURRENT FUTURES WITH THREADPOOLEXECUTOR INH A LOOP

import concurrent.futures
import time

start=time.perf_counter()

def do_something():
    print("Thread sleeping")
    time.sleep(1)
    return("done sleeping r")

l=[]
with concurrent.futures.ThreadPoolExecutor() as executor:
    for i in range(10):
        f1=executor.submit(do_something)
        l.append(f1)
    for t in l:
        print(t.result())

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')
```

```python
#task 5 - USING CONCUTRENT FUTURES USING THREADPOOLEXECUTOR IN LIST COMPHERENSION

import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds):
    print(f"sleeping for {seconds} sec")
    time.sleep(seconds)
    return "Done Sleeping"

with concurrent.futures.ThreadPoolExecutor() as executor:
    res=[executor.submit(do_something, 1) for i in range(10)]

    for f in concurrent.futures.as_completed(res):
        print(f.result())

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')

```

```python
#TASK 6 - PASSING TIME AS LIST TO EACH THREAD

import concurrent.futures
import time

start=time.perf_counter()
secs=[1,1.5,2,2.5,3,3.5,4,4.5,5,5.5]

def do_something(sec):
    print(f"sleeping for {sec} sec")
    time.sleep(sec)
    return "Done Sleeping"

with concurrent.futures.ThreadPoolExecutor() as executor:
     result=[executor.submit(do_something,sec) for sec in secs]

     for f in concurrent.futures.as_completed(result):
         print(f.result())

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')


```

```python
#TASK 7 - USING BUILT - IN MAP() METHOD

import concurrent.futures
import time

start=time.perf_counter()

def do_something(sec):
    print(f"sleeping for {sec} sec")
    time.sleep(sec)
    return "Done Sleeping"

with concurrent.futures.ThreadPoolExecutor() as executor:
    secs=[1,2,5,1,2,4]
    res=executor.map(do_something,secs)

for r in res:
    print(r)

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} second(s)')



    
```
